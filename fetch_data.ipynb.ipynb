```python
import yfinance as yf
import pandas as pd
tickers = ["AAPL","MSFT","GOOGL","TSLA","AMZN",
"META","NVDA","JPM","JNJ","V"]
stock_id_map = {
    "AAPL":1, "MSFT":2, "GOOGL":3, "TSLA":4, "AMZN":5,
    "META":6, "NVDA":7, "JPM":8,  "JNJ":9,  "V":10
}

for ticker in tickers:
    df = yf.download(ticker, start="2022-01-01", end="2024-12-31")
    df = df[["Open","Close","Volume"]].reset_index()
    df.columns = ["price_date","open_price","close_price","volume"]
    df["stock_id"] = stock_id_map[ticker]   # ← add this line
    df.to_csv(f"{ticker}.csv", index=False)
```
